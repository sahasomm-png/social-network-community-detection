# Social Network Community Detection and Influence Analysis

## 1. Project Overview

This project analyzes a real-world social network using graph-based techniques to identify communities of users and understand the structure of their connections.

The project uses the Facebook Social Network dataset from the Stanford Network Analysis Project (SNAP).

The network is represented as a graph where:
- Nodes represent users.
- Edges represent connections between users.

Community detection algorithms will be applied to identify groups of closely connected users. Graph centrality measures will also be used to identify influential users and users that act as bridges between different communities.

The project will compare multiple community detection algorithms and evaluate their performance using network-based metrics such as modularity, community size, and execution time.

## 2. Problem Statement

Social networks contain millions of users connected through relationships and interactions. Analyzing these connections can reveal hidden groups or communities within the network.

The objective of this project is to develop a graph-based system that analyzes social network connections and identifies communities of closely connected users.

The project will also analyze important users within the network and compare different community detection algorithms to determine which approach produces effective community structures.

## 3. Project Objectives

1. Load and preprocess the social network dataset.
2. Represent the social network as a graph.
3. Analyze the structural properties of the network.
4. Identify highly connected and influential users.
5. Detect communities using multiple graph algorithms.
6. Compare community detection algorithms.
7. Evaluate detected communities using modularity and other metrics.
8. Visualize the network and detected communities.
9. Extract meaningful insights from the network.
10. Discuss real-world applications of community detection.

```python
#Step 11 — Check Network Connectivity

# Connected components

components = list(nx.connected_components(G))

print("Number of connected components:", len(components))

component_sizes = sorted(
    [len(component) for component in components],
    reverse=True
)

print("Sizes of largest connected components:")
print(component_sizes[:10])
```

```python
#Step 12.1 — Degree Centrality

#Interpretation: A higher score means the user has more direct connections.

# Degree Centrality

degree_centrality = nx.degree_centrality(G)

top_degree = sorted(
    degree_centrality.items(),
    key=lambda x: x[1],
    reverse=True
)[:10]

print("Top 10 Users by Degree Centrality:\n")

for rank, (user, score) in enumerate(top_degree, 1):
    print(f"{rank}. User {user} → {score:.4f}")
```

Step 12.2 — Betweenness Centrality

This is particularly interesting for community detection.

Betweenness measures how often a user lies on the shortest paths between other users.

In simple terms:

A user with high betweenness can act as a bridge between different parts of the network.

```python
# Betweenness Centrality

betweenness_centrality = nx.betweenness_centrality(G)

top_betweenness = sorted(
    betweenness_centrality.items(),
    key=lambda x: x[1],
    reverse=True
)[:10]

print("Top 10 Users by Betweenness Centrality:\n")

for rank, (user, score) in enumerate(top_betweenness, 1):
    print(f"{rank}. User {user} → {score:.4f}")
```

Step 12.3 — Closeness Centrality

Closeness measures how quickly a user can reach other users through the network.

```python
# Closeness Centrality

closeness_centrality = nx.closeness_centrality(G)

top_closeness = sorted(
    closeness_centrality.items(),
    key=lambda x: x[1],
    reverse=True
)[:10]

print("Top 10 Users by Closeness Centrality:\n")

for rank, (user, score) in enumerate(top_closeness, 1):
    print(f"{rank}. User {user} → {score:.4f}")
```

Step 12.4 — PageRank

PageRank considers not only how many connections a user has, but also how important the users connecting to them are.

```python
# PageRank

pagerank = nx.pagerank(G)

top_pagerank = sorted(
    pagerank.items(),
    key=lambda x: x[1],
    reverse=True
)[:10]

print("Top 10 Users by PageRank:\n")

for rank, (user, score) in enumerate(top_pagerank, 1):
    print(f"{rank}. User {user} → {score:.6f}")
```

Step 12.5 — Create a Centrality Comparison Table

Now we'll combine everything into one DataFrame.

```python
import pandas as pd

centrality_df = pd.DataFrame({
    "User": list(G.nodes()),
    "Degree Centrality": [
        degree_centrality[node] for node in G.nodes()
    ],
    "Betweenness Centrality": [
        betweenness_centrality[node] for node in G.nodes()
    ],
    "Closeness Centrality": [
        closeness_centrality[node] for node in G.nodes()
    ],
    "PageRank": [
        pagerank[node] for node in G.nodes()
    ]
})

centrality_df.head()
```

What we're trying to discover

We now have four different definitions of an important user:

Measure     	                What it identifies
Degree Centrality	               Highly connected users
Betweenness Centrality	           Bridge users
Closeness Centrality	           Users who can reach others efficiently
PageRank	                       Users connected to other important users

This distinction is very useful in an interview.

For example, don't simply say:

"User X is the most important."

Instead:

"The definition of importance depends on the centrality measure. Degree centrality identifies highly connected users, whereas betweenness centrality identifies users that connect different parts of the network."

Step 13 — Louvain Community Detection

The goal is to automatically divide our social network into groups where users are more densely connected to each other than to users outside the group.

We'll use the Louvain algorithm first because it is widely used for large networks and gives us a modularity score to evaluate the quality of the communities.

13.1 Run Louvain

```python
from community import community_louvain

# Detect communities using Louvain algorithm
partition = community_louvain.best_partition(G)

print("Community detection completed!")
print("Number of users:", len(partition))
```

Step 13.2 — Count the Communities

Now let's see how many communities were detected.

```python
num_communities = len(set(partition.values()))

print("Number of communities detected:", num_communities)
```

Step 13.3 — Calculate Community Sizes

Now let's determine how many users belong to each community.

```python
from collections import Counter

community_sizes = Counter(partition.values())

print("Community sizes:\n")

for community, size in sorted(
    community_sizes.items(),
    key=lambda x: x[1],
    reverse=True
):
    print(f"Community {community}: {size} users")
```

Step 13.4 — Find the Largest Communities

Let's display the 10 largest communities.

```python
top_communities = community_sizes.most_common(10)

print("Top 10 Largest Communities:\n")

for rank, (community, size) in enumerate(top_communities, 1):
    print(f"{rank}. Community {community} → {size} users")
```

Step 13.5 — Calculate Modularity

This is very important for your project.

Modularity measures how well the network has been divided into communities.

Higher modularity generally means:

Users within the same community have relatively more connections to each other, while connections between different communities are relatively fewer.

```python
modularity_score = community_louvain.modularity(partition, G)

print("Louvain Modularity Score:", modularity_score)
```

For this type of network, you'll generally get a positive score indicating meaningful community structure.

Interview explanation

If an interviewer asks:

"How did you evaluate your community detection?"

You can answer:

"I used modularity to evaluate the quality of the detected communities. It measures whether there are more intra-community connections than would be expected under a suitable null model. A higher modularity indicates stronger community structure."

```python
# Calculate internal and external edges for each community

community_analysis = []

for community, size in community_sizes.items():

    nodes = {
        node for node in G.nodes()
        if partition[node] == community
    }

    internal_edges = 0
    external_edges = 0

    for node in nodes:
        for neighbor in G.neighbors(node):

            if partition[neighbor] == community:
                internal_edges += 1
            else:
                external_edges += 1

    # Each internal edge was counted twice
    internal_edges = internal_edges // 2

    community_analysis.append({
        "Community": community,
        "Users": size,
        "Internal Edges": internal_edges,
        "External Edges": external_edges
    })

community_analysis_df = pd.DataFrame(community_analysis)

community_analysis_df = community_analysis_df.sort_values(
    "Users",
    ascending=False
)

community_analysis_df.head(10)
```

```python
# Calculate internal connection ratio

community_analysis_df["Internal Ratio"] = (
    community_analysis_df["Internal Edges"] /
    (
        community_analysis_df["Internal Edges"] +
        community_analysis_df["External Edges"]
    )
)

community_analysis_df.head(10)
```

```python
# Communities with highest internal connection ratio

community_analysis_df.sort_values(
    by="Internal Ratio",
    ascending=False
).head(10)
```

```python
# Select top 10 largest communities

top_10_analysis = community_analysis_df.head(10)

plt.figure(figsize=(12, 6))

x = range(len(top_10_analysis))

plt.bar(
    x,
    top_10_analysis["Internal Edges"],
    label="Internal Edges"
)

plt.bar(
    x,
    top_10_analysis["External Edges"],
    bottom=top_10_analysis["Internal Edges"],
    label="External Edges"
)

plt.xlabel("Community")
plt.ylabel("Number of Edges")
plt.title("Internal vs External Connections of Top 10 Communities")

plt.xticks(
    x,
    top_10_analysis["Community"].astype(str)
)

plt.legend()
plt.show()
```

```python
from networkx.algorithms.community import girvan_newman
```

```python
import time

start_time = time.time()

gn_generator = girvan_newman(G_gn)

# Get the first useful community split
gn_communities = next(gn_generator)

end_time = time.time()

gn_time = end_time - start_time

print("Girvan-Newman completed!")
print("Number of communities:", len(gn_communities))
print("Execution time:", round(gn_time, 2), "seconds")
```

```python
# Calculate modularity of the Girvan-Newman communities

gn_modularity = nx.community.modularity(
    G_gn,
    gn_communities
)

print("Girvan-Newman Modularity:", round(gn_modularity, 4))
```

```python
# Analyze internal and external connections of each community

community_analysis = []

# Go through every detected community
for community, size in community_sizes.items():

    # Get all users belonging to this community
    nodes = {
        node for node in G.nodes()
        if partition[node] == community
    }

    internal_edges = 0
    external_edges = 0

    # Check the connections of every user
    for node in nodes:

        for neighbor in G.neighbors(node):

            # Connection stays inside the same community
            if partition[neighbor] == community:
                internal_edges += 1

            # Connection goes to another community
            else:
                external_edges += 1

    # Internal edges are counted twice, so divide by 2
    internal_edges = internal_edges // 2

    # Store the results
    community_analysis.append({
        "Community": community,
        "Users": size,
        "Internal Edges": internal_edges,
        "External Edges": external_edges
    })

# Convert results to DataFrame
community_analysis_df = pd.DataFrame(
    community_analysis
)

# Sort by community size
community_analysis_df = community_analysis_df.sort_values(
    "Users",
    ascending=False
).reset_index(drop=True)

# Display the results
community_analysis_df
```
